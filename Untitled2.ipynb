```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
# Set random seed for reproducible shower generations
np.random.seed(42)

# Number of secondary particles generated in the cascade
n_particles = 5000

# Atmospheric height (Primary hit occurs at ~20 km)
shower_height_km = 20.0

# Radial opening angle (determines horizontal spread of the shower cone)
spread_angle_rad = 0.08  # Radians
# Distance from shower axis at sea level (Exponential / NKG-like radial profile)
# Core is dense, while outer edges have sparse hits
radial_distances = np.random.exponential(scale=150.0, size=n_particles)  # meters

# Random azimuthal angles around the central axis (0 to 2*pi)
azimuth_angles = np.random.uniform(0, 2 * np.pi, size=n_particles)

# Convert polar coordinates (r, phi) to 2D Cartesian ground coordinates (X, Y)
x_ground = radial_distances * np.cos(azimuth_angles)
y_ground = radial_distances * np.sin(azimuth_angles)

# Simulate particle energy deposition (GeV) - core particles carry higher energy
energy_gev = (
    100.0 * np.exp(-radial_distances / 100.0)
    + np.random.exponential(scale=2.0, size=n_particles)
)
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(15, 6))

# --- PANEL 1: Ground Detector Array Footprint ---
scatter = ax1.scatter(
    x_ground,
    y_ground,
    c=energy_gev,
    cmap="plasma",
    s=12,
    alpha=0.7,
    edgecolors="none",
)

# Draw central shower core marker
ax1.scatter(
    [0],
    [0],
    color="cyan",
    marker="X",
    s=120,
    zorder=5,
    label="Shower Core Center",
)

# Set equal aspect ratio for realistic spatial geometry
ax1.set_aspect("equal")
ax1.set_xlim(-600, 600)
ax1.set_ylim(-600, 600)
ax1.set_title(
    "Cosmic Ray Air Shower Ground Footprint (Sea Level)",
    fontsize=13,
    pad=12,
)
ax1.set_xlabel("X Ground Position [meters]", fontsize=11)
ax1.set_ylabel("Y Ground Position [meters]", fontsize=11)
ax1.grid(True, linestyle=":", alpha=0.5)
ax1.legend(loc="upper right", fontsize=9)

# Add energy scale colorbar
cbar = plt.colorbar(scatter, ax=ax1)
cbar.set_label("Particle Energy [GeV]", fontsize=10)

# --- PANEL 2: Lateral Particle Density Distribution ---
# Bin radial distances to calculate particle density per ring area
bins = np.linspace(0, 500, 30)
counts, bin_edges = np.histogram(radial_distances, bins=bins)
bin_centers = 0.5 * (bin_edges[:-1] + bin_edges[1:])

# Calculate ring area for each bin: Area = pi * (r_outer^2 - r_inner^2)
ring_areas = np.pi * (bin_edges[1:] ** 2 - bin_edges[:-1] ** 2)
density = counts / ring_areas  # Particles per square meter

# Plot density vs radius on a semi-log scale
ax2.semilogy(
    bin_centers,
    density,
    "ro-",
    linewidth=2,
    markersize=5,
    label="Simulated Density Profile",
)

ax2.set_title(
    "Lateral Particle Density Profile $N(r)$", fontsize=13, pad=12
)
ax2.set_xlabel("Distance from Core $r$ [meters]", fontsize=11)
ax2.set_ylabel("Particle Density [Particles / $m^2$]", fontsize=11)
ax2.grid(True, which="both", linestyle=":", alpha=0.6)
ax2.legend(fontsize=10)

plt.tight_layout()
plt.show()
```
